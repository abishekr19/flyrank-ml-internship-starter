### Quick look at the data

The dataset contains 30,000 pages across 44 columns. A quick check shows that 6,068 pages have at least 100 impressions but zero clicks. More importantly, 4,166 pages have at least 250 impressions but still receive zero clicks. This suggests that a significant number of pages are getting visibility in search but are not generating clicks, making content improvement and search discoverability a useful area to investigate over the next 7 weeks.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Pages with 100+ impressions but zero clicks
zero_click_100 = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print("Total pages:", len(df))
print("Total columns:", len(df.columns))
print("Pages with 100+ impressions but 0 clicks:", len(zero_click_100))

# Pages with 250+ impressions but zero clicks
zero_click_250 = df[
    (df["impressions_90d"] >= 250) &
    (df["clicks_90d"] == 0)
]

print("Pages with 250+ impressions but 0 clicks:", len(zero_click_250))

# Position tier distribution
print("\nPosition tiers among 250+ impression / 0-click pages:")
print(zero_click_250["position_tier"].value_counts())
```
