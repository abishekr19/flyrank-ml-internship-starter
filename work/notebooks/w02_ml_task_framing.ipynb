### Success metric

The main evaluation metrics are **precision and recall**.

Recall is especially important for the declining class because missing a webpage that is actually declining could cause an important content-improvement opportunity to be missed.

Precision is also important because the content team should not spend time investigating too many pages that are unlikely to need attention.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML beats a fixed rule here

A fixed rule might say that every webpage with impressions below a certain threshold should be investigated.

However, search performance depends on multiple signals, such as impressions, clicks, click-through rate, and search position.

ML can learn combinations and patterns across these signals instead of relying on one manually chosen threshold.

For example, two webpages may have similar impressions but different clicks and search positions. A fixed rule may treat them similarly, while an ML model can learn that their situations are different.

The model output can help prioritize webpages for content investigation and improvement.

This is decision-support evidence, not a guarantee that changing a webpage will improve its search performance.
